```python
from safety_gridworlds_gymnasium.environments import *

import time
import gymnasium as gym
import pygame
import os
from gymnasium.wrappers import RecordEpisodeStatistics, RecordVideo

ENVIRONMENTS = [
    "safety_gridworlds/Base-v0",
    "safety_gridworlds/IslandNavigation-v0",
    "safety_gridworlds/Sokoban-v0",
    "safety_gridworlds/ConveyorBelt-v0",
]
active_env = ENVIRONMENTS[1]
```

```python
video_dir = os.path.join("exports", active_env)
os.makedirs(video_dir, exist_ok=True)

env = gym.make(active_env, render_mode="rgb_array")
env = RecordVideo(
    env,
    video_folder=video_dir,
    episode_trigger=lambda ep: True,  # record every episode
)
obs, info = env.reset()
```

```python
num_episodes = 5
max_steps_per_episode = 1000

episode_idx = 0
running = True


try:
    while running and episode_idx < num_episodes:
        obs, info = env.reset()
        total_reward = 0.0
        steps = 0

        for _ in range(max_steps_per_episode):
            action = env.action_space.sample()
            obs, reward, terminated, truncated, info = env.step(action)

            _ = env.render()

            total_reward += float(reward)
            steps += 1
            if terminated or truncated:
                break

        print(f"[{active_env}] Episode {episode_idx+1} finished: steps={steps}, return={total_reward:.3f}")
        episode_idx += 1

except KeyboardInterrupt:
    print("Interrupted by user.")

finally:
    env.close()

print(f"Videos saved to: {os.path.abspath(video_dir)}")
```
